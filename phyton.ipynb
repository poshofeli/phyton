```python
nombre = input("Ingrese su nombre: ")
disponible = input("Cuando dinero tienes? ")
gasto= input("Cual ha sido tu primer gasto hoy? ")
gasto2= input("Cual ha sido tu segundo gasto hoy? ")
gasto3= input("Cual ha sido tu tercer gasto hoy? ")
actual = float(disponible) - float(gasto) - float(gasto2) - float(gasto3)
porcentaje = float(actual*100)/float(disponible)
promedio = float(gasto)+float(gasto2)+float(gasto3)/(3)

print("Hola,"+nombre,"has gastado:"+str(float(gasto)+float(gasto2)+float(gasto3)),"tu saldo actual es:"+str(round(actual)))
print("Un %"+str(round(porcentaje,2)),"de su saldo inicial")
print("Un gasto promedio de:",str(round(promedio,3)))
```
